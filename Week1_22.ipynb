```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://th.bing.com/th/id/OIP.19xTfm0CQ9luUslYdTLNbwAAAA?w=135&h=150&c=7&r=0&o=5&dpr=1.1&pid=1.7"]

for url in urls:
  image = io.imread(url) #read gambar
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize gambar
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert warna ke RGB
  final_frame = cv.hconcat((image, image_2)) #concat gambar
  cv2_imshow(final_frame) #show gambar
  print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, "x", lebar)
cv2_imshow(image_2)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#buat garis horizonatal di TENGAH
for y in range (lebar):
  image_3[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for x in range(tinggi):
    image_3[x, int(lebar/2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))

cv2_imshow(final_frame)
```

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#garis diagonal dari kiri atas ke kanan bawah
for x in range(lebar):
    y = int((tinggi / lebar) * x)
    if y < tinggi:
        image_3[y, x] = [255, 255, 255]

#garis diagonal dari kiri bawah ke kanan atas
for x in range(lebar):
    y = int(tinggi - (tinggi / lebar) * x - 1)
    if y >= 0:
        image_3[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))

cv2_imshow(final_frame)
```

```python
#untuk menggambar garis horizontal
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

panjang_garis = 25

# Hitung posisi tengah gambar
y_tengah = int(tinggi / 2)
x_awal = int((lebar - panjang_garis) / 2)  # Posisi awal garis di bagian tengah
x_akhir = x_awal + panjang_garis  # Posisi akhir garis

for x in range(x_awal, x_akhir):
    image_3[y_tengah, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))

cv2_imshow(final_frame)
```

```python
lebar_kotak = 25  # Misalnya 50 piksel
tinggi_kotak = 25  # Misalnya 50 piksel

# Tentukan posisi kiri atas kotak secara acak
x_awal = 45
y_awal = 25

# Buat kotak dengan mengganti piksel dalam area tertentu menjadi putih
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)  # Copy gambar dengan konversi ke RGB
for y in range(y_awal, y_awal + tinggi_kotak):
    for x in range(x_awal, x_awal + lebar_kotak):
        image_3[y, x] = [255, 255, 255]  # Set piksel menjadi putih

# Tampilkan hasil
cv2_imshow(image_3)
```

*  Identifikasi kenapa menggunakan konversi warna BGR2RGB pada matplotlib dan tanpa konversi pada opencv.
*  Jawab: OpenCV menggunakan format BGR secara default untuk gambar, sehingga tidak memerlukan konversi warna saat menampilkan gambar dengan OpenCV. Matplotlib menggunakan format RGB, sehingga kamu perlu mengonversi gambar dari BGR ke RGB sebelum menampilkannya dengan Matplotlib.

```python
gambar = 'KTM_Putra Nindya Yuwana.jpg'

imagektm = io.imread(gambar) #read gambar
imagektm = cv.resize(imagektm, (0,0), fx=0.5, fy=0.5) #resize gambar
imagektm_2 = cv.cvtColor(imagektm, cv.COLOR_BGR2RGB) #convert warna ke RGB

tinggi_ktm = imagektm_2.shape[0]
lebar_ktm = imagektm_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi_ktm, "x", lebar_ktm)

cv2_imshow(imagektm_2) #show gambar
plt.imshow(imagektm) #showgambar Matplotlib
plt.axis('off')  #menyembunyikan sumbu
plt.title('Gambar Matplotlib')
plt.show()
```
